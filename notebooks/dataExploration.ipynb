```python
probCustIDs = ['HO-15230','HO-15231','HO-15232','HO-15233','HO-15234']
replacementID = 'HO-15230'

df[df['customer_id'].isin(probCustIDs)][['customer_id', 'customer_name', 'order_id', 'product_id']].sort_values('customer_id')
```

```python
#See which product names have mulitple product IDs associated with them
prodID_counts = df.groupby('product_name')['product_id'].nunique()
prodID_counts[prodID_counts > 1]
```

```python
#See which product IDs have multiple product names associated with them
prodName_counts = df.groupby('product_id')['product_name'].nunique()
prodName_counts[prodName_counts > 1].sort_values(ascending = False)
```

```python
problem_ids = prodName_counts[prodName_counts > 1].index

df.loc[df['product_id'].isin(problem_ids),['product_id', 'product_name']].drop_duplicates().sort_values('product_id').head(20)
```

```python
#Product IDs and product names have many to many relationships. Product ID cannot be used as a reliable unique identifier for products.
#Chose to leave Product data as is considering product ID will have little to no impact on the analysis.
```
